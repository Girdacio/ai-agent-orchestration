```python
# vamos rodar todos os testes unitários do projeto dentro de /test
import unittest

suite = unittest.defaultTestLoader.discover(start_dir='test', pattern='test*.py')
result = unittest.TextTestRunner(verbosity=2).run(suite)

if result.wasSuccessful():
    print("Todos os testes passaram.")
else:
    print("Há testes falhando.")
```
